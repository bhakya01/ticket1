3: Text Cleaning for Issue Descriptions

```python
def clean_issue_description(text):

    # Convert text to lowercase
    text = text.lower()

    # Replace slang / shorthand
    text = text.replace("ok", "okay")

    # Remove punctuation
    text = text.replace(".", "")
    text = text.replace(",", "")
    text = text.replace("!", "")
    text = text.replace("?", "")
    text = text.replace("-", "")

    text = " ".join(text.split())


    text = text.strip()

    return text


for i in range(len(ticket_data['Issue_Description'])):

    ticket_data['Issue_Description'][i] = clean_issue_description(
        ticket_data['Issue_Description'][i]
    )


print("\nCleaned Issue Descriptions")
print("-" * 50)

for i in range(len(ticket_data['Ticket_No'])):
    print(
        f"Ticket No: {ticket_data['Ticket_No'][i]} | "
        f"Issue: {ticket_data['Issue_Description'][i]}"
    )

```

2.Priority Analysis

```python
high_count = 0
medium_count = 0
low_count = 0

for priority in ticket_data['Priority']:
    if priority.lower() == "high":
        high_count += 1
    elif priority.lower() == "medium":
        medium_count += 1
    elif priority.lower() == "low":
        low_count += 1

print("\nPriority Analysis")
print("-" * 40)

print("High-priority tickets   :", high_count)
print("Medium-priority tickets :", medium_count)
print("Low-priority tickets    :", low_count)
```

3.Find the Ticket With the Longest Issue Description

```python

longest_index = 0
max_word_count = 0

for i in range(len(ticket_data['Issue_Description'])):
    word_count = len(ticket_data['Issue_Description'][i].split())

    if word_count > max_word_count:
        max_word_count = word_count
        longest_index = i


print("\nTicket With the Longest Issue Description")
print("-" * 50)

print("Ticket Number :", ticket_data['Ticket_No'][longest_index])
print("Customer Name :", ticket_data['Customer_Name'][longest_index])
print("Cleaned Issue :", ticket_data['Issue_Description'][longest_index])
print("Word Count    :", max_word_count)
```

4. Extract Unique Words Used

```python
unique_words = set()

for description in ticket_data['Issue_Description']:
    words = description.split()

    for word in words:
        unique_words.add(word)

sorted_words = sorted(unique_words)

print("\nUnique Words Analysis")
print("-" * 50)

print("Count of Unique Words:", len(sorted_words))

print("\nUnique Words (Sorted):")
print(sorted_words)
```
